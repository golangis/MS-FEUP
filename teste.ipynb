```python
def get_info(data, column):
    # Get the row with the highest "Total Traffic Input"
    max_traffic = data[column].idxmax()
    print("Max:\n", data.loc[max_traffic])

    # Get the row with the lowest "Total Traffic Input"
    min_traffic = data[column].idxmin()
    print("\nMinimum:\n", data.loc[min_traffic])

    # Get the average "Total Traffic Input"
    avg_traffic = data[column].mean()
    print("\nAverage:", avg_traffic)

    # Get the median "Total Traffic Input"
    median_traffic = data[column].median()
    print("Median:", median_traffic)

    # Get the standard deviation of "Total Traffic Input"
    std_traffic = data[column].std()
    print("Standard Deviation:", std_traffic)
```

```python
import pandas as pd

# Load the data
auto = pd.read_csv('auto_traffic_model_results.csv')
allow_priority = pd.read_csv('allow_priority_traffic_model_results.csv')
priority_score = pd.read_csv('priority_score_traffic_model_results.csv')
multi = pd.read_csv('multi_agent_traffic_model_results.csv')

# Add a column indicating the model for each dataset
auto['Model'] = 'Auto'
allow_priority['Model'] = 'Allow Priority'
priority_score['Model'] = 'Priority Score'
multi['Model'] = 'Multi-Agent'

# Combine the datasets into a single DataFrame
combined_data = pd.concat([auto, allow_priority, priority_score, multi])

print('Total Accidents')
# Calculate the statistics for the 'Accidents' column
accident_stats = combined_data.groupby('Model')['Total Accidents'].agg(
    Average='mean',
    Median='median',
    Standard_Deviation='std'
).reset_index()

# Display the table with statistics
print(accident_stats)


print('Total Traffic Output')
# Calculate the statistics for the 'Accidents' column
accident_stats = combined_data.groupby('Model')['Total Traffic Output'].agg(
    Average='mean',
    Median='median',
    Standard_Deviation='std'
).reset_index()

# Display the table with statistics
print(accident_stats)


print('Total Traffic Input')
# Calculate the statistics for the 'Accidents' column
accident_stats = combined_data.groupby('Model')['Total Traffic Input'].agg(
    Average='mean',
    Median='median',
    Standard_Deviation='std'
).reset_index()

# Display the table with statistics
print(accident_stats)


print('Average Total Waiting Time')
# Calculate the statistics for the 'Accidents' column
accident_stats = combined_data.groupby('Model')['Average Total Waiting Time'].agg(
    Average='mean',
    Median='median',
    Standard_Deviation='std'
).reset_index()

# Display the table with statistics
print(accident_stats)

print('Priority Average Waiting Time')
# Calculate the statistics for the 'Accidents' column
accident_stats = combined_data.groupby('Model')['Priority Average Waiting Time'].agg(
    Average='mean',
    Median='median',
    Standard_Deviation='std'
).reset_index()

# Display the table with statistics
print(accident_stats)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
auto = pd.read_csv('auto_traffic_model_results.csv')
allow_priority = pd.read_csv('allow_priority_traffic_model_results.csv')
priority_score = pd.read_csv('priority_score_traffic_model_results.csv')
multi = pd.read_csv('multi_agent_traffic_model_results.csv')

# Add a column indicating the model for each dataset
auto['Model'] = 'Auto'
allow_priority['Model'] = 'Allow Priority'
priority_score['Model'] = 'Priority Score'
multi['Model'] = 'Multi-Agent'

# Combine the datasets into a single DataFrame
combined_data = pd.concat([auto, allow_priority, priority_score, multi])

# Specify the column to analyze
column = 'Total Traffic Input'

# Bin the values into intervals of 500
bin_width = 500
combined_data['Traffic Bin'] = pd.cut(
    combined_data[column],
    bins=range(0, int(combined_data[column].max()) + bin_width, bin_width),
    right=False,
    labels=[f"{i}-{i+bin_width-1}" for i in range(0, int(combined_data[column].max()), bin_width)]
)

# Count the occurrences of each bin per model
value_counts = combined_data.groupby(['Model', 'Traffic Bin']).size().reset_index(name='Count')

# Plot the number of occurrences of each bin for each model
plt.figure(figsize=(14, 8))
bar_plot = sns.barplot(data=value_counts, x='Traffic Bin', y='Count', hue='Model', palette='tab10')

# Annotate the bars with their count values
for bar in bar_plot.containers:
    bar_plot.bar_label(bar, fmt='%.0f', label_type='edge', padding=3)

plt.xlabel('Traffic Input Bins')
plt.ylabel('Count')
plt.title('Count of Traffic Input Bins Across Models')
plt.legend(title='Traffic Models')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
auto = pd.read_csv('auto_traffic_model_results.csv')
allow_priority = pd.read_csv('allow_priority_traffic_model_results.csv')
priority_score = pd.read_csv('priority_score_traffic_model_results.csv')
multi = pd.read_csv('multi_agent_traffic_model_results.csv')

# Add a column indicating the model for each dataset
auto['Model'] = 'Auto'
allow_priority['Model'] = 'Allow Priority'
priority_score['Model'] = 'Priority Score'
multi['Model'] = 'Multi-Agent'

# Combine the datasets into a single DataFrame
combined_data = pd.concat([auto, allow_priority, priority_score, multi])

# Specify the column to analyze
column = 'Total Accidents'

# Separate rows with 0 accidents
zero_accidents = combined_data[combined_data[column] == 0]
non_zero_accidents = combined_data[combined_data[column] > 0]

# Bin the values for non-zero accidents
bin_width = 2
bin_edges = [1] + list(range(2, int(non_zero_accidents[column].max()) + bin_width + 1, bin_width))
non_zero_accidents['Traffic Bin'] = pd.cut(
    non_zero_accidents[column],
    bins=bin_edges,
    right=False,
    labels=[f"{i}-{i+bin_width-1}" for i in bin_edges[:-1]]
)

# Count the occurrences of bins and 0 accidents per model
binned_counts = non_zero_accidents.groupby(['Model', 'Traffic Bin']).size().reset_index(name='Count')
zero_accidents_counts = zero_accidents.groupby('Model').size().reset_index(name='Count')
zero_accidents_counts['Traffic Bin'] = '0 Accidents'

# Combine the zero and binned data
all_counts = pd.concat([zero_accidents_counts, binned_counts])

# Ensure "0 Accidents" appears first in the plot
bin_order = ['0 Accidents'] + [f"{i}-{i+bin_width-1}" for i in bin_edges[:-1]]
all_counts['Traffic Bin'] = pd.Categorical(all_counts['Traffic Bin'], categories=bin_order, ordered=True)

# Plot the data
plt.figure(figsize=(14, 8))
sns.barplot(data=all_counts, x='Traffic Bin', y='Count', hue='Model', palette='tab10')

# Customize the plot
plt.xlabel('Total Accidents Bins')
plt.ylabel('Count')
plt.title('Count of Accidents Across Models')
plt.legend(title='Traffic Models')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
auto = pd.read_csv('auto_traffic_model_results.csv')
allow_priority = pd.read_csv('allow_priority_traffic_model_results.csv')
priority_score = pd.read_csv('priority_score_traffic_model_results.csv')
multi = pd.read_csv('multi_agent_traffic_model_results.csv')

# Add a column indicating the model for each dataset
auto['Model'] = 'Auto'
allow_priority['Model'] = 'Allow Priority'
priority_score['Model'] = 'Priority Score'
multi['Model'] = 'Multi-Agent'

# Combine the datasets into a single DataFrame
combined_data = pd.concat([auto, allow_priority, priority_score, multi])

# Specify the column to analyze
column = 'Average Total Waiting Time'

# Bin the values into intervals of 500 up to 520, and create a separate bin for values higher than 520
bin_width = 20
max_value = 520
upper_limit = 1000000  # The upper limit for the values greater than 520

# Create bins from 0 to 520 and then one bin for values greater than 520
bins = list(range(0, max_value + bin_width, bin_width)) + [upper_limit]
bin_labels = [f"{i}-{i+bin_width-1}" for i in range(0, max_value, bin_width)] + [f"{max_value}" + "+"]

combined_data['Traffic Bin'] = pd.cut(
    combined_data[column],
    bins=bins,
    right=False,
    labels=bin_labels
)

# Count the occurrences of each bin per model
value_counts = combined_data.groupby(['Model', 'Traffic Bin']).size().reset_index(name='Count')

# Plot the number of occurrences of each bin for each model
plt.figure(figsize=(14, 8))
bar_plot = sns.barplot(data=value_counts, x='Traffic Bin', y='Count', hue='Model', palette='tab10')

# Annotate the bars with their count values
for bar in bar_plot.containers:
    bar_plot.bar_label(bar, fmt='%.0f', label_type='edge', padding=3)

plt.xlabel('Average Waiting Time Bins')
plt.ylabel('Count')
plt.title('Count of Average Waiting Time Across Models')
plt.legend(title='Traffic Models')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
